# 1. Black-Scholes

```python
from scipy.stats import norm
from scipy.optimize import newton

#Define call option price for given parameters
def black_scholes_call(S, K, r, sigma, T):
    """
    Calculate Black-Scholes for european call option
    """
    d_plus = (np.log(S/K) + (r + sigma**2/2)*T)/(sigma*np.sqrt(T))
    d_minus = d_plus - sigma*np.sqrt(T)
    
    call = S*norm.cdf(d_plus) - K*np.exp(-r*T)*norm.cdf(d_minus)
    return call

#Calculate IV for given parameters
def implied_volatility(C,S,K, r, T, sigma_init=0.2):
    """
    Calculating implied volatility using Newton-Raphson method
    
    Parameters:
    C (float): Market price of call option
    S
    K
    r
    T
    sigma_init
    """
    def objective(sigma):
        return (black_scholes_call(S,K,r,sigma,T) - C)**2
    
    implied_vol = newton(objective,sigma_init)
    return implied_vol





```
