## Lifted Heston

In this notebook we will explore the lifted Heston model (LHM) and illustrate it's convergence to the rough Heston model (RHM):

We first look at the convergence of $K^n \to K$

```python
import scipy as sp

H = 0.1 # We take H = 0.1 as deomnstrated by Gatheral


def K(t):       # Our fractional kernel
    return (H * (t ** (H - 0.5))) / sp.special.gamma(H + 0.5)

def r(n):       # Parameter used for the approximation of the fractional kernel
    return n^{1/n}
    
def K(t,n):  # Our fractional kernel approximation
    return (H * (t ** (H - 0.5))) / sp.special.gamma(H + 0.5) * r(n)
    





    
```
